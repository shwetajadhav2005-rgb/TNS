```python
import numpy as np
# ---- THE DATA ----

student_names = np.array(["Aisha", "Ben", "Carla", "Dev", "Eli", "Farah"])   # six students
subject_names = np.array(["Maths", "Physics", "English"])                    # three subjects

scores = np.array([[78, 65, 82],    # Aisha's three marks - one per subject, in the order above
                   [55, 71, 60],    # Ben's
                   [91, 88, 79],    # Carla's
                   [42, 38, 55],    # Dev's
                   [67, 74, 71],    # Eli's
                   [83, 97, 68]])   # Farah's

print(scores)                       # the whole table
print("shape:", scores.shape)
#print total marks of each student , average marks of each student. print each student name , total and average marks
total = np.sum(scores , axis=1)
print(total)

average = np.mean(scores , axis=1)
print(average)

# Print name, total and average
for i in range(len(student_names)):
    print(student_names[i], "Total:", total[i], "Average:", round(average[i], 1))
```

Draw a bar chart of eact students physics mark.Sort the table by physics first and give chartall three lables

```python

student_names = ["Aisha", "Ben", "Carla", "Dev", "Eli", "Farah"]
physics = [65, 72, 78, 82, 88, 63]

# Combine names and marks, then sort by Physics marks
data = sorted(zip(physics, student_names))

# Separate them again
sorted_physics = [x[0] for x in data]
sorted_names = [x[1] for x in data]

# Bar chart
plt.bar(sorted_names, sorted_physics)

plt.title("Students' Physics Marks")
plt.xlabel("Student Names")
plt.ylabel("Physics Marks")

plt.show()
```

Break it down:

sorted_physics → variable that stores sorted marks.
[ ] → creates a list.
for x in data → takes each item from data, one by one.
x → represents one pair, for example (63, "Farah").
x[0] → gets the first value from the pair.

Remember Python indexing:

x[0] → first value → Physics mark
x[1] → second value → Student name

Therefore:

sorted_physics

becomes:

[63, 65, 72, 78, 82, 88]
3. Get only student names
sorted_names = [x[1] for x in data]

Here:

sorted_names → variable for student names.
for x in data → goes through every pair.
x[1] → gets the second value, which is the name.

Result:

["Farah", "Aisha", "Ben", "Carla", "Dev", "Eli"]
4. Draw the bar chart
plt.bar(sorted_names, sorted_physics)
plt → matplotlib.pyplot, which is used for drawing graphs.
bar() → creates a bar chart.
sorted_names → goes on the X-axis.
sorted_physics → goes on the Y-axis.

So:

X-axis → Student Names
Y-axis → Physics Marks
5. Give the chart a title
plt.title("Students' Physics Marks")
plt.title() → gives a title to the chart.
"Students' Physics Marks" → the title we want to display.
6. Give X-axis a label
plt.xlabel("Student Names")
xlabel() → means X-axis label.
"Student Names" → appears below the X-axis.
7. Give Y-axis a label
plt.ylabel("Physics Marks")
ylabel() → means Y-axis label.
"Physics Marks" → appears beside the Y-axis.
